```python
LANG2S = {
    "svo": {
        "eng": "verb",
        "zho": "verb",
        "rus": "verb",
        "tur": "object",
        "jap": "object",
    },
    "modal-verbs": {
        "eng": "modal",
        "zho": "modal",
        "tur": "verb",
        "ger": "modal"
    },
    "noun-adj": {
        "eng": "adj",
        "ger": "adj",
        "rus": "adj",
        "zho": "adj",
        "ned": "adj",
        "fre": "noun",
        "ita": "noun",
        "vie": "noun",
    },
    "noun-adj-complete": {
        "eng": "adj",
        "ger": "adj",
        "rus": "adj",
        "zho": "adj",
        "ned": "adj",
        "fre": "noun",
        "ita": "noun",
        "vie": "noun",
    }
}

ROLE_COLORS = {"src": "blue", "tgt": "red", "eng": "green"}
```

```python
# Two shapes: one for "matches expected POS of tgt", one for "matches expected POS of src"
ROLE_SHAPES = {"tgt": "circle", "src": "square", "mismatch": "x"}

def get_shape(pos, lang_role, src_lang, tgt_lang, experiment):
    """Return shape based on whether pos matches the expected POS of src or tgt."""
    expected_tgt = LANG2S.get(experiment, {}).get(tgt_lang)
    expected_src = LANG2S.get(experiment, {}).get(src_lang)
    if pos == expected_tgt:
        return ROLE_SHAPES["tgt"]
    if pos == expected_src:
        return ROLE_SHAPES["src"]
    return ROLE_SHAPES["mismatch"]


for (experiment, model_short), exp_df in all_data.groupby(['experiment', 'model_short']):

    src_langs = sorted(exp_df['src_lang'].unique())[:4]
    tgt_langs = sorted(exp_df['tgt_lang'].unique())[:4]

    n_rows, n_cols = len(tgt_langs), len(src_langs)

    fig = make_subplots(
        rows=n_rows, cols=n_cols,
        row_titles=tgt_langs,
        column_titles=src_langs,
        shared_xaxes=True,
        shared_yaxes=True,
    )

    for col_i, src_lang in enumerate(src_langs, start=1):
        for row_i, tgt_lang in enumerate(tgt_langs, start=1):

            ROLE_LABELS = {
                "src": "L<sup>src</sup>",
                "tgt": "L<sup>tgt</sup>",
                "eng": "English",
            }

            cell_df = exp_df[
                (exp_df['src_lang'] == src_lang) &
                (exp_df['tgt_lang'] == tgt_lang)
            ]
            if cell_df.empty:
                continue
            
            def lang_to_role(lang, src_lang, tgt_lang):
                if lang == "eng":
                    return "eng"
                if lang == tgt_lang:
                    return "tgt"
                if lang == src_lang:
                    return "src"
                return None

            for token_type in sorted(cell_df['token_type'].unique()):
                parts     = token_type.split('-')
                pos, lang = parts[0], parts[-1]
                role      = lang_to_role(lang, src_lang, tgt_lang)
                color     = ROLE_COLORS[role] if role else "grey"
                label     = ROLE_LABELS[role] if role else lang
                symbol    = get_shape(pos, role, src_lang, tgt_lang, experiment)

                avg_data = (
                    cell_df[cell_df['token_type'] == token_type]
                    .groupby('layer')['prob'].mean()
                    .reset_index()
                )

                fig.add_trace(
                    go.Scatter(
                        x=avg_data['layer'],
                        y=avg_data['prob'],
                        mode='lines+markers',
                        line=dict(color=color),
                        marker=dict(color=color, symbol=symbol),
                        legendgroup=label,
                        showlegend=False,
                        name=label,
                    ),
                    row=row_i, col=col_i,
                )

    # Dummy traces — role legend (lines only, no markers)
    for role, label in ROLE_LABELS.items():
        fig.add_trace(go.Scatter(
            x=[None], y=[None],
            mode='lines',
            name=label,
            line=dict(color=ROLE_COLORS[role], width=3),
            legendgroup=label,
            legendgrouptitle=dict(text="Language") if role == "src" else None,
            showlegend=True,
        ))

    # Dummy traces — marker shape legend (black markers only, no lines)
    shape_legend = {
        "S<sup>tgt</sup>": ROLE_SHAPES["tgt"],
        "S<sup>src</sup>": ROLE_SHAPES["src"],
        "other":           ROLE_SHAPES["mismatch"],
    }
    for i, (label, symbol) in enumerate(shape_legend.items()):
        fig.add_trace(go.Scatter(
            x=[None], y=[None],
            mode='markers',
            name=label,
            marker=dict(color='black', symbol=symbol, size=8),
            legendgrouptitle=dict(text="Expected PoS") if i == 0 else None,
            legendgroup="pos",
            showlegend=True,
        ))

    fig.update_layout(
        title_text=f'[{experiment}] Probability Evolution — {model_short}',
        height=300 * n_rows + 100,
        width=350 * n_cols + 150,
        legend=dict(font=dict(size=24)),
    )

    # Axis titles only on bottom row and leftmost column
    fig.update_xaxes(title_text='Layer', row=n_rows)
    fig.update_yaxes(title_text='Probability', col=1)

    out_path = os.path.join(output_dir, f'prob_evolution_{experiment}_{model_short}_grid-mini.html')
    fig.write_html(out_path)
    pdf_path = os.path.join(output_dir, f'prob_evolution_{experiment}_{model_short}_grid-mini.pdf')
    fig.write_image(pdf_path, format='pdf')
    print(f"Saved PDF: {pdf_path}")
    print(f"Saved: {out_path}")
    fig.show()
```
